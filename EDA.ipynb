```python
kolom_kategorik = df.select_dtypes(include=['object', 'str']).columns

for kolom in kolom_kategorik:
    print(kolom)
    print(df[kolom].unique())
    print()

#Yoseva Faurensia Sitohang_202431195
```

```python
for kolom in df.select_dtypes(include=['object', 'str']).columns:
    if df[kolom].isna().sum() > 0:
        df[kolom] = df[kolom].fillna(df[kolom].mode()[0])

for kolom in df.select_dtypes(include='number').columns:
    if df[kolom].isna().sum() > 0:
        df[kolom] = df[kolom].fillna(df[kolom].mean())

df.isna().sum()

#Yoseva Faurensia Sitohang_202431195
```

```python
print("Jumlah data duplikat:", df.duplicated().sum())

df = df.drop_duplicates()

df.info()

#Yoseva Faurensia Sitohang_202431195
```

```python
jumlah_gender = df['gender'].value_counts()

jumlah_gender.plot(kind='bar')

plt.title('Jumlah Siswa Berdasarkan Gender')
plt.xlabel('Gender')
plt.ylabel('Jumlah Siswa')
plt.xticks(rotation=0)
plt.show()

#Yoseva Faurensia Sitohang_202431195
```
