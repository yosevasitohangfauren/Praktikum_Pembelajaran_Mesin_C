```python
import requests
import csv
import pandas as pd
from bs4 import BeautifulSoup

# Yoseva Faurensia Sitohang_202431195
```

```python
robots_url = "https://pokemondb.net/robots.txt"

robots_response = requests.get(robots_url, timeout=30)

print("URL robots.txt:", robots_url)
print("Status Code:", robots_response.status_code)
print(robots_response.text)

# Yoseva Faurensia Sitohang_202431195
```

```python
url = "https://pokemondb.net/pokedex/all"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers, timeout=30)

print("URL halaman:", url)
print("Status Code:", response.status_code)

# Yoseva Faurensia Sitohang_202431195
```

```python
soup = BeautifulSoup(response.text, "html.parser")

table = soup.find("table", id="pokedex")

if table:
    rows = table.find("tbody").find_all("tr")
    print("Jumlah data:", len(rows))
else:
    rows = []
    print("Tabel Pokémon tidak ditemukan.")

# Yoseva Faurensia Sitohang_202431195
```

```python
result = []

for row in rows:
    cells = row.find_all("td")

    if len(cells) >= 10:
        data = {
            "nomor": cells[0].get_text(strip=True),
            "nama": cells[1].get_text(strip=True),
            "tipe": cells[2].get_text(" ", strip=True),
            "total": cells[3].get_text(strip=True),
            "hp": cells[4].get_text(strip=True),
            "attack": cells[5].get_text(strip=True),
            "defense": cells[6].get_text(strip=True),
            "sp_attack": cells[7].get_text(strip=True),
            "sp_defense": cells[8].get_text(strip=True),
            "speed": cells[9].get_text(strip=True)
        }

        result.append(data)

print("Jumlah data berhasil diekstrak:", len(result))

# Yoseva Faurensia Sitohang_202431195
```

```python
for data in result[:10]:
    print(data)

# Yoseva Faurensia Sitohang_202431195
```

```python
nama_file = "pokemon_dataset.csv"

fieldnames = [
    "nomor", "nama", "tipe", "total", "hp",
    "attack", "defense", "sp_attack", "sp_defense", "speed"
]

with open(nama_file, "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(result)

print("File berhasil dibuat:", nama_file)

# Yoseva Faurensia Sitohang_202431195
```

```python
df = pd.read_csv("pokemon_dataset.csv")

print("Ukuran dataset:", df.shape)
print("Jumlah baris:", len(df))
print("Nama kolom:", df.columns.tolist())

display(df.head())

# Yoseva Faurensia Sitohang_202431195
```

```python
kolom_angka = [
    "nomor", "total", "hp", "attack",
    "defense", "sp_attack", "sp_defense", "speed"
]

for kolom in kolom_angka:
    df[kolom] = pd.to_numeric(df[kolom], errors="coerce")

print("Tipe data setiap kolom:")
print(df.dtypes)

print("\nJumlah data kosong:")
print(df.isna().sum())

# Yoseva Faurensia Sitohang_202431195
```

```python
df.to_csv("pokemon_dataset.csv", index=False)

print("Dataset berhasil disimpan kembali.")

# Yoseva Faurensia Sitohang_202431195
```
